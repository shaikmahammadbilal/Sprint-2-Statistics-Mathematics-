## Section A: Theoretical Questions & Answers

### Q1. Explain the difference between Mean and Median.
**Answer:**
- **Mean** is the simple average of all numbers. You calculate it by adding all numbers together and dividing by the total count. It is highly affected by extreme values or outliers.
- **Median** is the exact middle point of the data when arranged in ascending or descending order. It splits the data into two equal halves. It is not affected by outliers and represents typical center data much better when extreme values are present.

### Q2. What is Variance?
**Answer:**
Variance measures how far spread out the numbers in your dataset are from the mean value. It calculates the average of the squared differences from the Mean. A high variance means data points are spread widely apart, while a low variance means they are tightly clustered around the average.

### Q3. What is Standard Deviation?
**Answer:**
Standard Deviation is simply the square root of the Variance. It tells you the average distance of data points from the Mean. The main benefit is that it is expressed in the exact same unit as the original data, making it much easier to understand compared to variance.

### Q4. Explain Bayes' Theorem with an example.
**Answer:**
Bayes' Theorem calculates the probability of an event happening based on prior knowledge or conditions related to that event. 

**Formula:** $P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$

**Simple Example:** Suppose a doctor is checking if a patient has a specific rare disease. 
1. Before any test, there is a general baseline chance (prior probability) that anyone has the disease.
2. The patient takes a lab test, and it comes back positive.
3. Bayes' theorem combines the accuracy rate of the test with the baseline disease rate to give the actual final probability that this specific person really has the disease.

### Q5. Difference between Population and Sample.
**Answer:**
- **Population** refers to the entire complete collection of elements or individuals that you want to study (e.g., all registered voters in India).
- **Sample** is a small, manageable subgroup selected directly from that population (e.g., surveying 10,000 voters across different Indian states). We study the sample to draw conclusions about the whole population.

### Q6. Explain Correlation and Covariance.
**Answer:**
- **Covariance** tells us the direction of a relationship between two variables. If positive, they move together; if negative, they move in opposite directions. However, it does not tell us the strength because its value depends on the units of the data.
- **Correlation** is the standardized version of covariance. It tells us both the direction and the exact strength of the linear relationship. Its value is strictly locked between -1 and +1, making it highly useful for direct comparisons.

### Q7. What is Skewness?
**Answer:**
Skewness is a measure of the asymmetry or lack of symmetry in a data distribution curve. 
- **Positive Skew:** The tail on the right side of the distribution is longer or fatter (mean is greater than median).
- **Negative Skew:** The tail on the left side is longer or fatter (mean is less than median).
- **Zero Skew:** The data is perfectly balanced and symmetric on both sides like a normal bell curve.

### Q8. Explain IQR.
**Answer:**
IQR stands for Interquartile Range. It measures statistical dispersion by calculating the difference between the 75th percentile (Q3) and the 25th percentile (Q1). Essentially, it focuses strictly on the middle 50% of your data distribution, completely ignoring extreme outliers on either side.

### Q9. What is a p-value?
**Answer:**
A p-value (probability value) measures the strength of evidence against a Null Hypothesis. It tells you how likely it is to observe your data by total random chance if the null hypothesis is completely true. A small p-value (usually $\le 0.05$) means the result is highly unlikely to be a fluke, leading you to reject the null hypothesis in favor of a statistically significant finding.

### Q10. Difference between Type I and Type II Error.
**Answer:**
- **Type I Error (False Positive):** This happens when you reject a true null hypothesis. For example, a medical test says a patient has a disease when they are completely healthy.
- **Type II Error (False Negative):** This happens when you fail to reject a false null hypothesis. For example, a medical test misses a disease and tells a sick patient that they are completely healthy.

### Exercise 11: Write Python code to calculate Variance from scratch without external libraries.

```python
def calculate_variance(data):
    # 1. Find the mean
    n = len(data)
    if n == 0:
        return 0
    mean_val = sum(data) / n
    
    # 2. Sum up squared differences
    squared_diff_sum = sum((x - mean_val) ** 2 for x in data)
    
    # 3. Divide by n (population variance) or n-1 (sample variance)
    # Using sample variance (n - 1) as a standard rule
    variance = squared_diff_sum / (n - 1) if n > 1 else 0
    return variance

# Sample Test Data
scores = [85, 90, 78, 92, 88]
print("Sample Dataset:", scores)
print("Calculated Sample Variance:", calculate_variance(scores))
```

### Exercise 12: Detect Outliers using the IQR Method.

```python
import numpy as np

def detect_outliers_iqr(data):
    # 1. Calculate Q1 and Q3 using numpy percentiles
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)
    
    # 2. Calculate IQR
    iqr = q3 - q1
    
    # 3. Calculate lower and upper boundaries
    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)
    
    # 4. Extract outliers
    outliers = [x for x in data if x < lower_bound or x > upper_bound]
    return outliers, lower_bound, upper_bound

# Sample Data with extreme values (12 and 150)
dataset = [12, 45, 48, 50, 52, 53, 55, 58, 60, 150]
outliers_found, low, high = detect_outliers_iqr(dataset)

print("Dataset:", dataset)
print(f"Bounds: Lower = {low}, Upper = {high}")
print("Detected Outliers:", outliers_found)
```

### Exercise 13: Calculate Correlation between two variables.

```python
import pandas as pd

# 1. Create a dummy data dictionary representing study hours vs exam marks
data = {
    'Study_Hours': [2, 4, 5, 7, 8, 10],
    'Exam_Marks': [45, 55, 60, 75, 80, 95]
}

# 2. Convert dictionary to a pandas DataFrame
df = pd.DataFrame(data)

# 3. Compute Pearson correlation matrix
correlation_matrix = df.corr(method='pearson')

# 4. Extract specific correlation value between the two variables
corr_value = correlation_matrix.loc['Study_Hours', 'Exam_Marks']

print("Data Table:\n", df)
print(f"\nPearson Correlation Coefficient: {corr_value:.4f}")
```

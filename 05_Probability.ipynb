## 1. Introduction to Probability

### Definition
Probability tells us how likely an event is to happen.

Probability is always between **0 and 1**:
- `0` = impossible
- `1` = certain
- `0.5` = 50% chance

### Formula

P(A) = Number of favorable outcomes / Total number of possible outcomes

### Numerical Example
A fair coin has 2 outcomes: Head and Tail.

P(Head) = 1 / 2 = 0.5 = 50%

### Business Example
A company may estimate the probability that a customer will buy a product.

### AI/ML Example
A classification model may predict that a customer has an 80% probability of buying.

### Interpretation
Probability describes chance; it is not a guarantee.

```python
# Probability of getting Head
favorable_outcomes = 1
total_outcomes = 2

probability_head = favorable_outcomes / total_outcomes

print("Probability of Head:", probability_head)
print("Percentage:", probability_head * 100, "%")
```

## 2. Sample Space

### Definition
The **sample space** is the collection of all possible outcomes of an experiment.

It is usually represented by `S`.

### Numerical Example
For one die roll:

S = {1, 2, 3, 4, 5, 6}

### Business Example
A customer survey could have the sample space:
`{Satisfied, Neutral, Unsatisfied}`.

### AI/ML Example
For a three-class model:
`{Cat, Dog, Bird}` can represent possible classes.

```python
# Sample space of a six-sided die
sample_space = {1, 2, 3, 4, 5, 6}

print("Sample Space:", sample_space)
print("Total outcomes:", len(sample_space))

# Probability of getting an even number
even_outcomes = {2, 4, 6}
probability_even = len(even_outcomes) / len(sample_space)

print("P(Even):", probability_even)
```

## 3. Events

### Definition
An **event** is one or more outcomes from the sample space that we are interested in.

### Example
For a die:

S = {1, 2, 3, 4, 5, 6}

Event A = getting an even number:

A = {2, 4, 6}

P(A) = 3 / 6 = 0.5

### Types
- Simple event: one outcome
- Compound event: more than one outcome
- Certain event: probability = 1
- Impossible event: probability = 0

```python
sample_space = {1, 2, 3, 4, 5, 6}
event_even = {2, 4, 6}

probability = len(event_even) / len(sample_space)

print("Event:", event_even)
print("Probability:", probability)
```

### Business Example
Event = “A customer purchases a product.”

### AI/ML Example
Event = “The model predicts that an email is spam.”

### Interpretation
An event is the outcome, or group of outcomes, that we want to study.

## 4. Independent Events

### Definition
Two events are **independent** if one event does not change the probability of the other.

### Formula

P(A and B) = P(A) × P(B)

### Numerical Example
Flip a coin and roll a die.

P(Head) = 1/2

P(6) = 1/6

P(Head and 6) = 1/2 × 1/6 = 1/12

### Business Example
Two unrelated random processes can sometimes be modeled as independent.

### AI/ML Example
Some probabilistic models make independence assumptions to simplify calculations.

### Interpretation
Independent events can happen together. Independence means one does not change the probability of the other.

```python
p_head = 1 / 2
p_six = 1 / 6

p_both = p_head * p_six

print("P(Head):", p_head)
print("P(6):", p_six)
print("P(Head and 6):", p_both)
```

## 5. Dependent Events

### Definition
Two events are **dependent** when one event changes the probability of the other.

### Numerical Example
A bag contains 3 red balls and 2 blue balls.

First draw:
P(Red) = 3/5

If a red ball is removed, there are 2 red balls among 4 balls.

Second draw:
P(Red | Red first) = 2/4 = 1/2

### Business Example
When a limited-stock product is sold, the chance that the next customer gets that product can change.

### AI/ML Example
Real-world features or events can depend on previous conditions.

### Interpretation
If an earlier event changes the situation, later probability can change.

```python
red = 3
blue = 2
total = red + blue

p_red_first = red / total

# Remove one red ball
red_after = red - 1
total_after = total - 1

p_red_second = red_after / total_after

print("P(Red first):", p_red_first)
print("P(Red second | Red first):", p_red_second)
```

## 6. Conditional Probability

### Definition
Conditional probability is the probability of an event when we already know that another event has happened.

### Formula

P(A | B) = P(A and B) / P(B)

### Numerical Example
In a class:
- 18 students know Python.
- 12 students know both Python and SQL.

P(SQL | Python) = 12/18 = 0.667 ≈ 66.7%

### Business Example
Probability that a customer buys a product **given that** they clicked an advertisement.

### AI/ML Example
Conditional probability is used in classification and probabilistic models.

### Interpretation
The condition changes the group we are considering.

```python
python_students = 18
python_and_sql_students = 12

p_sql_given_python = python_and_sql_students / python_students

print("P(SQL | Python):", p_sql_given_python)
print("Percentage:", p_sql_given_python * 100, "%")
```

## 7. Bayes' Theorem

### Definition
Bayes' Theorem updates a probability when new evidence is available.

### Formula

P(A | B) = [P(B | A) × P(A)] / P(B)

Where:
- P(A) = prior probability
- P(B | A) = likelihood
- P(A | B) = posterior probability

### Numerical Example
Suppose:
- 1% of customers are fraudulent.
- The system flags 90% of fraudulent customers.
- It falsely flags 5% of genuine customers.

We want P(Fraud | Flag).

P(Flag) = (0.90 × 0.01) + (0.05 × 0.99)

P(Fraud | Flag) ≈ 15.38%

### Business Example
Fraud detection and risk assessment.

### AI/ML Example
Bayesian reasoning is used in Naive Bayes and other probabilistic models.

### Interpretation
When an event is rare, even a fairly accurate test can have a much lower positive predictive probability than expected.

```python
p_fraud = 0.01
p_flag_given_fraud = 0.90
p_flag_given_not_fraud = 0.05
p_not_fraud = 1 - p_fraud

p_flag = (
    p_flag_given_fraud * p_fraud
    + p_flag_given_not_fraud * p_not_fraud
)

p_fraud_given_flag = p_flag_given_fraud * p_fraud / p_flag

print("P(Flag):", round(p_flag, 4))
print("P(Fraud | Flag):", round(p_fraud_given_flag, 4))
print("Percentage:", round(p_fraud_given_flag * 100, 2), "%")
```

## 8. Probability Rules

### Rule 1: Complement Rule

P(not A) = 1 - P(A)

Example:
If P(Rain) = 0.30,

P(No Rain) = 1 - 0.30 = 0.70

### Rule 2: Addition Rule

P(A or B) = P(A) + P(B) - P(A and B)

If A and B are mutually exclusive:

P(A or B) = P(A) + P(B)

### Rule 3: Multiplication Rule

P(A and B) = P(A) × P(B | A)

For independent events:

P(A and B) = P(A) × P(B)

```python
# Complement rule
p_rain = 0.30
print("P(No Rain):", 1 - p_rain)

# Addition rule
p_a = 0.40
p_b = 0.50
p_a_and_b = 0.20

print("P(A or B):", p_a + p_b - p_a_and_b)

# Multiplication rule
p_a = 0.60
p_b_given_a = 0.40

print("P(A and B):", p_a * p_b_given_a)
```

### Business Example
Probability rules can help calculate the chance that a customer buys product A or product B.

### AI/ML Example
Probability rules are used when combining evidence and calculating likelihoods.

### Interpretation
Always check whether events overlap and whether they are independent or dependent.

## 9. Random Variables

### Definition
A **random variable** is a variable whose value depends on the outcome of a random experiment.

### Types

**Discrete random variable**
- Takes countable values.
- Example: number of heads in 3 coin tosses.

**Continuous random variable**
- Can take any value in an interval.
- Example: customer waiting time.

### Business Example
Number of products sold in one hour is discrete.

Customer waiting time is continuous.

### AI/ML Example
Random variables are the foundation of probability distributions and statistical models.

```python
# Discrete random variable:
# Number of heads in 3 coin tosses

outcomes = ["TTT", "TTH", "THT", "THH",
            "HTT", "HTH", "HHT", "HHH"]

number_of_heads = [outcome.count("H") for outcome in outcomes]

print("Outcomes:", outcomes)
print("Number of heads:", number_of_heads)
```

## 10. Expected Value

### Definition
Expected value is the long-run average outcome of a random variable.

### Formula

E(X) = Σ [x × P(X = x)]

### Numerical Example

| Outcome | Probability |
|---|---:|
| ₹0 | 0.5 |
| ₹100 | 0.3 |
| ₹200 | 0.2 |

Expected value:

E(X) = (0 × 0.5) + (100 × 0.3) + (200 × 0.2)

E(X) = ₹70

### Business Example
A company can calculate expected revenue, profit, cost, or loss.

### AI/ML Example
Expected value is used in decision-making, risk analysis, and reinforcement learning.

### Interpretation
Expected value is a long-run average. It does not mean every individual result will be ₹70.

```python
import numpy as np

outcomes = np.array([0, 100, 200])
probabilities = np.array([0.5, 0.3, 0.2])

expected_value = np.sum(outcomes * probabilities)

print("Expected Value: ₹", expected_value)
```

## 11. Probability Distribution Example

A probability distribution assigns probabilities to possible values of a random variable.

For a valid discrete probability distribution:
- Every probability is between 0 and 1.
- All probabilities add up to 1.

```python
import matplotlib.pyplot as plt
import numpy as np

values = np.array([1, 2, 3, 4])
probabilities = np.array([0.10, 0.20, 0.30, 0.40])

print("Sum of probabilities:", probabilities.sum())

plt.bar(values, probabilities)
plt.title("Discrete Probability Distribution")
plt.xlabel("Random Variable X")
plt.ylabel("P(X)")
plt.show()

expected_value = np.sum(values * probabilities)
print("Expected Value:", expected_value)
```

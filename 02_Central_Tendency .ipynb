# Measures of Central Tendency (Simple Notes)

This notebook explains the standard ways to find the **center point** (average) of your data. We use simple language, real-world Indian business examples, and Python code to understand these concepts clearly.

--- 
## 1. Arithmetic Mean (Simple Average)

### Definition
The **Mean** is the most common average. You just add up all the numbers in a dataset and divide the total by the count of numbers.

### Formula
$$\mu = \frac{\sum_{i=1}^{n} x_i}{n}$$

Where:
* $\sum x_i$ = Sum of all values
* $n$ = Total number of values

### Numerical Example
Suppose a small tea stall sells 5 cups, 10 cups, 15 cups, 20 cups, and 25 cups of tea over 5 hours.  
$$\text{Mean} = \frac{5 + 10 + 15 + 20 + 25}{5} = \frac{75}{5} = 15 \text{ cups}$$

### Python Implementation

```python
import numpy as np
import pandas as pd
from scipy import stats

tea_sales = [5, 10, 15, 20, 25]

# Using NumPy
mean_val = np.mean(tea_sales)
print(f"Mean using NumPy: {mean_val}")
```

### Business Example
A local Kirana shop owner tracks his daily sales revenue for a week: `[₹1200, ₹1500, ₹1100, ₹1300, ₹1900, ₹2100, ₹1400]`. Finding the mean daily sales helps him calculate how much stock to buy for the next month.

### AI/ML Example
When preparing data to train a machine learning model, some columns might have missing data (NaN). In **Mean Imputation**, we fill these empty boxes with the average value of that column so the model can learn properly.

### Interpretation
The mean gives a quick picture of the center, but it has one big problem: **it gets easily ruined by extreme values (outliers)**. For example, if a celebrity visits the tea stall and leaves a ₹10,000 tip, the mean will shoot up, making it look like the stall is highly profitable every day, which is not true.

## 2. Median (Middle Value)

### Definition
The **Median** is the exact middle number when you arrange your data in order (from smallest to biggest). If you have an even number of values, you take the average of the two middle numbers.

### Formula
If $n$ is **Odd**:
$$\text{Median} = X_{\left(\frac{n+1}{2}\right)}$$

If $n$ is **Even**:
$$\text{Median} = \frac{X_{\left(\frac{n}{2}\right)} + X_{\left(\frac{n}{2} + 1\right)}}{2}$$

### Numerical Example
* **Odd count:** Data = `[10, 20, 35, 40, 50]` $\rightarrow$ Middle value is **35**.
* **Even count:** Data = `[10, 20, 35, 40, 50, 60]` $\rightarrow$ Average of 35 and 40 = **37.5**.

### Python Implementation

```python
salaries = [15000, 18000, 22000, 25000, 30000, 95000]
median_val = np.median(salaries)
print(f"Median Value: {median_val}")
```

### Business Example
Look at apartment prices in a Mumbai neighborhood. Five flats cost ₹60 Lakhs, ₹65 Lakhs, ₹70 Lakhs, ₹75 Lakhs, and one luxury penthouse costs ₹15 Crores (₹1500 Lakhs). The mean price is skewed heavily to ~₹354 Lakhs, but the median is a sensible ₹70 Lakhs. Real estate agents always prefer the median.

### AI/ML Example
In computer vision, **Median Filtering** is used to clean up digital images. If an image has "salt-and-pepper" noise (random black and white dots), replacing each pixel with the median value of surrounding pixels wipes out the noise perfectly without blurring the edges.

### Interpretation
The median splits your data into two perfect halves: 50% values are below it and 50% are above it. It is highly **robust**, meaning it doesn't care about extreme outliers.

## 3. Mode (Most Frequent Value)

### Definition
The **Mode** is simply the value that appears most frequently in your dataset. A dataset can have one mode, multiple modes, or no mode at all.

### Formula
$$\text{Mode} = \text{Value with the highest frequency } (f_i)$$

### Numerical Example
Dataset of shoe sizes sold: `[6, 7, 8, 8, 8, 9, 10]`  
Size 8 appears 3 times. So, **Mode = 8**.

### Python Implementation

```python
shoe_sizes = [6, 7, 8, 8, 8, 9, 10]
mode_result = stats.mode(shoe_sizes, keepdims=True)
print(f"Mode: {mode_result.mode[0]} (Appeared {mode_result.count[0]} times)")
```

### Business Example
A clothing store manager reviews the T-shirt sizes bought by customers: `[S, M, M, L, L, L, L, XL]`. The mode size is `L`. The manager uses this information to order more 'Large' stock from the factory.

### AI/ML Example
In the **K-Nearest Neighbors (KNN)** classification algorithm, a new data point is labeled based on what the majority of its closest neighbors are. This "majority voting" is nothing but finding the mode.

### Interpretation
Mode is excellent for non-numerical data (like favorite color or brand name) where you cannot calculate a math average. It tells you what is the most popular choice.

## 4. Weighted Mean

### Definition
Sometimes, not all numbers carry equal importance. In **Weighted Mean**, we multiply each value by a specific weight that shows how important it is before taking the average.

### Formula
$$\bar{x}_w = \frac{\sum_{i=1}^{n} w_i x_i}{\sum_{i=1}^{n} w_i}$$

Where:
* $x_i$ = Values
* $w_i$ = Weights

### Numerical Example
A student gets marks in a college term:  
* Assignments (Weight = 20%): 80 marks  
* Mid-term Exam (Weight = 30%): 90 marks  
* Final Exam (Weight = 50%): 70 marks  
$$\bar{x}_w = \frac{(80 \times 0.2) + (90 \times 0.3) + (70 \times 0.5)}{0.2 + 0.3 + 0.5} = \frac{16 + 27 + 35}{1.0} = 78 \text{ marks}$$

### Python Implementation

```python
marks = [80, 90, 70]
weights = [0.2, 0.3, 0.5]
weighted_avg = np.average(marks, weights=weights)
print(f"Weighted Mean Grade: {weighted_avg}")
```

### Business Example
An investor buys shares of a company multiple times: 100 shares at ₹50, and later 400 shares at ₹60. To know the true average cost price per share, the investor calculates the weighted mean based on the volume bought at each price.

### AI/ML Example
Modern AI systems like ChatGPT use **Attention Mechanisms** (in Transformers). The neural network assigns dynamic weights to words in a sentence, giving more importance (higher weight) to context words that help understand meaning.

### Interpretation
Weighted mean is crucial when certain components naturally dominate others, making a regular flat average inaccurate.

## 5. Geometric Mean

### Definition
The **Geometric Mean** multiplies all numbers together and then takes the $n$-th root (where $n$ is the total count of numbers). It is strictly used when numbers scale multiplicatively, like percentages or growth rates.

### Formula
$$GM = \sqrt[n]{x_1 \cdot x_2 \dots x_n} = \left(\prod_{i=1}^{n} x_i\right)^{\frac{1}{n}}$$

### Numerical Example
Your investment grows by a multiplier of 2 in Year 1, and 8 in Year 2.  
$$GM = \sqrt{2 \times 8} = \sqrt{16} = 4$$

### Python Implementation

```python
from scipy.stats import gmean

growth_multipliers = [2, 8]
geo_mean = gmean(growth_multipliers)
print(f"Geometric Mean: {geo_mean}")
```

### Business Example
When calculating **Compound Annual Growth Rate (CAGR)** of a business's revenue over 5 years, using a simple arithmetic mean will give an over-inflated, wrong percentage. The Geometric Mean must be used to get the true compounding growth rate.

### AI/ML Example
When optimizing complex multi-objective models or evaluating metrics where different features are on completely different scales, the Geometric Mean helps normalize everything fairly without letting one large range dominate the calculation.

### Interpretation
The geometric mean will always be less than or equal to the regular mean. It acts as the accurate center for growth trends. *Note: It cannot handle zero or negative numbers.*

## 6. Harmonic Mean

### Definition
The **Harmonic Mean** is calculated by dividing the total count of items by the sum of the reciprocals ($1/x$) of each item. It is the gold standard for averaging **rates, ratios, or speeds**.

### Formula
$$HM = \frac{n}{\sum_{i=1}^{n} \frac{1}{x_i}}$$

### Numerical Example
You drive to an office at 40 km/h and return via the exact same route at 60 km/h. What is the average speed?  
$$HM = \frac{2}{\frac{1}{40} + \frac{1}{60}} = \frac{2}{\frac{3 + 2}{120}} = \frac{2 \times 120}{5} = 48 \text{ km/h}$$

### Python Implementation

```python
from scipy.stats import hmean

speeds = [40, 60]
harm_mean = hmean(speeds)
print(f"Harmonic Mean Speed: {harm_mean}")
```

### Business Example
A factory manager wants to evaluate machine productivity. Machine X takes 10 minutes per unit and Machine Y takes 20 minutes per unit. The Harmonic Mean helps find the true average system time taken per batch.

### AI/ML Example
In classification models, we check **Precision** (how accurate predictions are) and **Recall** (how many targets were caught). The famous **F1-Score** is the Harmonic Mean of precision and recall. It ensures that if even one of them is terrible, the overall score drops heavily.

### Interpretation
The harmonic mean gives more weight to smaller values. This means a single lazy machine or poor rate cannot hide behind other high-performing rates.

## 7. Cheat Sheet: How to Pick the Right Average?

Choosing the right measure depends entirely on your data distribution shape and data type. Use this simple guide:

| Data Type / Distribution Shape | Best Measure to Use | Why? |
| :--- | :--- | :--- |
| **Symmetric / Bell Curve** (No Outliers) | **Arithmetic Mean** | It naturally sits in the exact center when data is perfectly balanced. |
| **Skewed Data / Has Outliers** (e.g., Salaries, House Prices) | **Median** | Outliers do not affect it; it safely shows what a typical data point looks like. |
| **Categorical Data** (e.g., Best Colors, T-shirt Sizes, Product Names) | **Mode** | You cannot calculate math averages on words. Mode finds the popular item. |
| **Items with Asymmetric Importance** (e.g., Marks with different weights) | **Weighted Mean** | Ensures critical items contribute proportionally to the final score. |
| **Growth, Returns & Percentages** (e.g., CAGR, Investment Returns) | **Geometric Mean** | Accurately tracks multiplicative scaling without compounding inflation. |
| **Rates, Speeds & Ratios** (e.g., km/h, Unit production time, F1-Score) | **Harmonic Mean** | Correctly weights proportional rates sharing a fixed distance/workload. |

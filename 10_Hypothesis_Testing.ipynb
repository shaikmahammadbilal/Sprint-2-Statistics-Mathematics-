### Learning goal
Hypothesis testing helps us use sample data to make decisions about a population.

### Simple idea
We start with a claim, collect sample data, and use statistics to decide whether the evidence is strong enough to reject the starting assumption.

## 1. Null Hypothesis (H₀)

### Definition
The **null hypothesis** is the starting assumption that there is **no effect, no difference, or no relationship**.

### Real-world analogy
A company says: "Our new website does not change the average customer order value."

That statement is the starting assumption.

### Small numerical example
Old average order value = ₹500  
New sample average = ₹530

We might define:

**H₀:** New average order value = ₹500

### Business example
A company wants to know whether a new advertising campaign increases sales.

### AI/ML use case
Suppose we compare two model versions. H₀ can state that their average performance is the same.

### How the example demonstrates the concept
The null hypothesis gives us a baseline. We do not automatically assume that the observed ₹30 difference is meaningful.

## 2. Alternative Hypothesis (H₁ / Hₐ)

### Definition
The **alternative hypothesis** represents the effect or difference we want to find evidence for.

### Example
If the company wants to know whether the new website changes average order value:

**H₀:** Mean order value = ₹500  
**H₁:** Mean order value ≠ ₹500

### Real-world analogy
H₀ says "nothing changed." H₁ says "something changed."

### Business example
Testing whether a new pricing strategy changes average revenue.

### AI/ML use case
Testing whether a new model has a different average error from the old model.

### Important
A statistical test provides evidence for or against H₀. It does not prove H₁ with absolute certainty.

## 3. p-value

### Definition
The **p-value** is the probability, assuming the null hypothesis is true, of observing results at least as extreme as the result obtained.

### Simple interpretation
A small p-value means the observed result would be unusual if H₀ were true.

### Common rule
If `p < 0.05`, we often reject H₀.

### Real-world analogy
Imagine a fair coin. If you flip it 10 times and get heads 10 times, you would question the assumption that the coin is fair.

### Small numerical example
Suppose a test produces:

`p-value = 0.02`

Because `0.02 < 0.05`, we reject H₀ at the 5% significance level.

### Business example
A/B testing shows `p = 0.02`. There is evidence that the two versions perform differently.

### AI/ML use case
Compare model errors or experimental performance between two approaches.

### Important
A p-value is **not** the probability that H₀ is true.

## 4. Confidence Interval

### Definition
A confidence interval gives a range of plausible values for a population parameter based on sample data.

A 95% confidence interval is commonly used.

### Real-world analogy
Instead of saying "the average delivery time is exactly 30 minutes," we might estimate a reasonable range such as 28–32 minutes.

### Small numerical example
Sample mean = 30 minutes  
95% confidence interval = 28 to 32 minutes

This means the interval is an estimate of plausible values for the population mean.

### Business example
A company estimates average customer waiting time as 28–32 minutes with a 95% confidence procedure.

### AI/ML use case
Confidence intervals can be used to report uncertainty around estimated model metrics or business parameters.

### Important interpretation
In frequentist statistics, a 95% confidence procedure means that if we repeatedly sampled and constructed intervals in the same way, about 95% of those intervals would contain the true parameter.

```python
import numpy as np
from scipy import stats

data = np.array([28, 30, 31, 29, 32, 30, 27, 33, 31, 29])

mean = data.mean()
confidence_level = 0.95
sem = stats.sem(data)

interval = stats.t.interval(
    confidence_level,
    df=len(data) - 1,
    loc=mean,
    scale=sem
)

print("Sample mean:", mean)
print("95% confidence interval:", interval)
```

### How the example demonstrates the concept
The sample gives an estimate of the average waiting time, while the interval shows uncertainty around that estimate.

## 5. Significance Level (α)

### Definition
The **significance level**, written as alpha (α), is the threshold chosen before testing for deciding when evidence against H₀ is considered statistically significant.

Common choices:
- α = 0.05
- α = 0.01
- α = 0.10

### Decision rule
- If `p < α` → reject H₀
- If `p ≥ α` → fail to reject H₀

### Real-world analogy
Think of α as a decision boundary. You decide how strong the evidence must be before changing your conclusion.

### Business example
A pharmaceutical company may choose α = 0.01 for a particularly strict analysis.

### AI/ML use case
Choose α before evaluating an experiment comparing model performance.

## 6. Type I Error

### Definition
A **Type I error** occurs when we reject H₀ even though H₀ is actually true.

This is also called a **false positive**.

### Real-world analogy
A security alarm says there is a burglar when there is actually no burglar.

### Small numerical example
H₀: "The new campaign does not improve sales."

If we conclude that the campaign improves sales when it actually does not, that is a Type I error.

### Business example
A company launches an expensive campaign because a statistical test incorrectly indicated an improvement.

### AI/ML use case
A model evaluation experiment may incorrectly conclude that a new model is better when it actually is not.

## 7. Type II Error

### Definition
A **Type II error** occurs when we fail to reject H₀ even though H₀ is false.

This is also called a **false negative**.

### Real-world analogy
A security alarm fails to detect a real burglar.

### Small numerical example
H₀: "The new campaign does not improve sales."

If the campaign really improves sales but the test fails to detect the improvement, that is a Type II error.

### Business example
A company may stop a useful product experiment because the sample was too small to detect the improvement.

### AI/ML use case
A model improvement may be real but not statistically detected because the experiment has insufficient data.

## 8. One-Tailed Test

### Definition
A one-tailed test checks for an effect in **one specified direction**.

### Example
A company wants to know whether a new process **increases** productivity.

**H₀:** Mean productivity ≤ 100 units  
**H₁:** Mean productivity > 100 units

### Real-world analogy
You are checking only one side of a road: "Did productivity go up?"

### Business example
Testing whether a new sales strategy increases average sales.

### AI/ML use case
Testing a pre-specified claim that a new model improves a metric.

### Important
The direction should be chosen before looking at the results.

```python
from scipy.stats import ttest_1samp

productivity = [104, 106, 102, 108, 105, 107, 103, 109]
stat, two_sided_p = ttest_1samp(productivity, popmean=100)

# For a greater-than alternative
one_sided_p = two_sided_p / 2 if stat > 0 else 1 - two_sided_p / 2

print("t-statistic:", stat)
print("One-tailed p-value:", one_sided_p)
```

## 9. Two-Tailed Test

### Definition
A two-tailed test checks whether a parameter is **different in either direction**.

### Example

**H₀:** Mean productivity = 100  
**H₁:** Mean productivity ≠ 100

The result could be significantly higher or significantly lower.

### Real-world analogy
Instead of asking "Did the temperature increase?", you ask "Did the temperature change?"

### Business example
Testing whether a new website changes average order value, whether higher or lower.

### AI/ML use case
Testing whether two models have different average performance without specifying a direction in advance.

```python
from scipy.stats import ttest_1samp

productivity = [104, 106, 102, 108, 105, 107, 103, 109]

stat, p_value = ttest_1samp(productivity, popmean=100)

print("t-statistic:", stat)
print("Two-tailed p-value:", p_value)
```

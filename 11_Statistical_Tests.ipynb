## 1. T-Test

### Definition
A t-test checks whether a mean is significantly different from a reference value or whether two means are significantly different.

### Real-world analogy
Suppose a teacher wants to know whether a new teaching method changes average marks. We compare the average marks before and after the method.

### Small numerical example
Group A: `70, 72, 68, 75, 71`
Group B: `80, 78, 82, 79, 81`

The averages look different. A t-test helps determine whether the difference is statistically significant.

### Assumptions
- Numerical/continuous outcome.
- Independent observations for an independent t-test.
- Approximately normal data for the usual small-sample version.
- For the classic pooled two-sample t-test, equal variances are assumed; Welch's t-test does not require equal variances.

### When to use
Use when comparing means, especially with smaller samples.

### When not to use
Do not use it for categorical outcomes or strongly unsuitable data. For paired measurements, use a paired t-test.

### Business example
Compare average sales before and after a new sales strategy.

### AI/ML use case
Compare average model errors across two experimental conditions.

```python
from scipy.stats import ttest_ind

group_a = [70, 72, 68, 75, 71]
group_b = [80, 78, 82, 79, 81]

stat, p_value = ttest_ind(group_a, group_b, equal_var=False)

print("t-statistic:", stat)
print("p-value:", p_value)
print("Significant at 5%:", p_value < 0.05)
```

### Interpretation
If `p < 0.05`, we reject the null hypothesis of equal population means and conclude that the sample provides evidence of a difference.

## 2. Z-Test

### Definition
A z-test compares a sample mean with a population/reference mean when the population standard deviation is known or the large-sample normal approximation is appropriate.

### Real-world analogy
A factory says its bottles contain an average of 500 ml. A quality team checks whether the production mean is consistent with 500 ml.

### Assumptions
- Independent observations.
- Numerical outcome.
- Population standard deviation is known for the classical one-sample z-test, or a large-sample approximation is justified.

### When to use
Use for large samples or when population standard deviation is known.

### When not to use
For a small sample with unknown population standard deviation, a t-test is usually preferred.

### Business example
Test whether average delivery time differs from a company target.

### AI/ML use case
Compare a large-sample model metric against a known benchmark when the assumptions support a z-based procedure.

```python
import numpy as np
from scipy.stats import norm

sample = np.array([498, 501, 500, 497, 502, 499, 503, 500, 501, 498])
population_mean = 500
population_std = 3

z = (sample.mean() - population_mean) / (population_std / np.sqrt(len(sample)))
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Z-statistic:", z)
print("p-value:", p_value)
```

## 3. Chi-Square Test

### Definition
A chi-square test is commonly used for categorical data. It can test association between categorical variables or compare observed counts with expected counts.

### Real-world analogy
A retailer wants to know whether product preference depends on customer age group.

### Small numerical example
Observed category counts are compared with expected counts. Large differences contribute more to the chi-square statistic.

### Assumptions
- Data are counts/frequencies.
- Observations are independent.
- Expected cell counts should generally be sufficiently large for the usual approximation.

### When to use
Use for categorical variables and frequency/count data.

### When not to use
Do not use it directly for continuous measurements as if they were categories.

### Business example
Test whether payment method preference is associated with customer segment.

### AI/ML use case
Evaluate associations between categorical features during exploratory analysis and feature selection.

```python
from scipy.stats import chi2_contingency

table = [
    [30, 20, 10],
    [20, 25, 15]
]

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square:", chi2)
print("p-value:", p_value)
print("Degrees of freedom:", dof)
print("Expected counts:")
print(expected)
```

### Interpretation
If `p < 0.05`, there is evidence of an association between the categorical variables in the contingency table.

Note: association does not automatically mean causation.

## 4. ANOVA

### Definition
ANOVA (Analysis of Variance) tests whether the means of **three or more groups** are equal.

### Real-world analogy
A company tests three advertising campaigns. ANOVA asks whether average sales differ across the three campaigns.

### Small numerical example
Campaign A: `10, 12, 11`
Campaign B: `15, 16, 14`
Campaign C: `20, 19, 21`

The group means are visibly different.

### Assumptions
- Independent observations.
- Approximately normal residuals within groups.
- Similar variances across groups for the standard one-way ANOVA.

### When to use
Use for comparing means across 3+ independent groups.

### When not to use
Do not use standard one-way ANOVA when assumptions are badly violated; consider alternatives such as Welch ANOVA or a nonparametric method depending on the situation.

### Business example
Compare average sales for three pricing strategies.

### AI/ML use case
Compare average model performance across multiple model-development conditions.

```python
from scipy.stats import f_oneway

group_a = [10, 12, 11, 13, 12]
group_b = [15, 16, 14, 17, 16]
group_c = [20, 19, 21, 22, 20]

f_stat, p_value = f_oneway(group_a, group_b, group_c)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

### Interpretation
If `p < 0.05`, at least one group mean differs. ANOVA itself does not tell us which groups differ; post-hoc tests can be used for that.

## 5. Mann-Whitney U Test

### Definition
The Mann-Whitney U test is a nonparametric test used to compare two independent groups.

### Real-world analogy
Instead of comparing exact exam marks directly, imagine ranking all students and asking whether one group's ranks tend to be higher.

### Assumptions
- Two independent groups.
- Outcome is at least ordinal.
- Observations are independent.
- Interpretation as a location/median difference is simplest when group distributions have similar shapes.

### When to use
Useful when normality is doubtful or data are ordinal/skewed.

### When not to use
Do not treat it as a universal replacement for a t-test; it tests distribution/rank differences, not always simply means or medians.

### Business example
Compare customer satisfaction ratings between two branches.

### AI/ML use case
Compare two model error distributions when normality assumptions are questionable.

```python
from scipy.stats import mannwhitneyu

group_a = [3, 4, 4, 5, 3, 4]
group_b = [5, 5, 4, 6, 5, 4]

stat, p_value = mannwhitneyu(group_a, group_b, alternative="two-sided")

print("U-statistic:", stat)
print("p-value:", p_value)
```

## 6. Shapiro-Wilk Test

### Definition
The Shapiro-Wilk test checks whether sample data are consistent with a normal distribution.

### Real-world analogy
Before using a method that works best with bell-shaped data, we perform a quick check to see whether the data look compatible with that shape.

### Assumptions
- Independent observations.
- Continuous data are typical.
- It is intended as a normality test, not as proof that data are perfectly normal.

### When to use
Use as one piece of evidence when checking normality, especially with small/moderate samples.

### When not to use
Do not rely only on the p-value for very large datasets. Also inspect histograms, Q-Q plots, and domain knowledge.

### Business example
Check whether customer wait times are approximately normal before choosing a statistical method.

### AI/ML use case
Check model residuals during regression diagnostics.

```python
from scipy.stats import shapiro

data = [10, 12, 11, 13, 12, 14, 9, 11, 13, 12]

stat, p_value = shapiro(data)

print("Shapiro-Wilk statistic:", stat)
print("p-value:", p_value)
```

### Interpretation
A common decision rule is:
- `p >= 0.05`: insufficient evidence to reject normality.
- `p < 0.05`: evidence against normality.

This is not proof of normality. Always combine the test with plots and context.

## 7. Quick Test Selection Guide

| Question | Possible test |
|---|---|
| Compare one mean with a reference | One-sample t-test / z-test |
| Compare two independent means | Independent t-test |
| Compare two paired measurements | Paired t-test |
| Compare 3+ independent means | ANOVA |
| Compare two independent non-normal/ordinal groups | Mann-Whitney U |
| Association between categorical variables | Chi-square |
| Check normality | Shapiro-Wilk |

### Final lesson
A statistical test is not just a Python function. Choose it based on the question, data type, design, assumptions, and business context.

> **Simple idea:** An outlier is a data value that is unusually far from most other values.

## 1. What are Outliers?

An **outlier** is an observation that is very different from the other observations.

### Real-world analogy
Imagine 9 students score between 60 and 80 marks, but one student scores 5 marks. The score 5 is unusual compared with the rest.

### Small numerical example
Data: `10, 11, 12, 13, 14, 50`

Here, **50** is much larger than the other values, so it may be an outlier.

### Business example
A shop normally receives 20–50 orders per day, but one day it receives 2,000 orders. This may be a genuine event, a promotion, or a data-entry error.

### AI/ML use case
Outlier detection is used for fraud detection, network intrusion detection, unusual transactions, and sensor monitoring.

### Important
An outlier is not automatically an error. First investigate why it occurred.

## 2. Types of Outliers

### 1. Point outlier
One individual value is unusual.

**Example:** `10, 11, 12, 13, 100`

### 2. Contextual outlier
A value is unusual only in a particular context.

**Example:** A temperature of 35°C may be normal in summer but unusual for a winter day.

### 3. Collective outlier
A group of observations is unusual together.

**Example:** A sudden sequence of thousands of failed login attempts can indicate an attack.

### Real-world analogy
A single loud noise may be unusual. A whole series of loud noises at midnight is unusual as a pattern.

## 3. Detecting Outliers

Common approaches:
1. Visual inspection
2. IQR method
3. Z-Score method
4. Domain knowledge

The best method depends on the data distribution and the business problem.

## 4. IQR Method

**IQR = Q3 − Q1**

The common outlier boundaries are:

- Lower boundary = `Q1 − 1.5 × IQR`
- Upper boundary = `Q3 + 1.5 × IQR`

Values outside these boundaries are flagged as potential outliers.

### Real-world analogy
Think of Q1 and Q3 as the lower and upper edges of the main group. The IQR method checks whether a value is too far beyond that group.

```python
import numpy as np

data = np.array([10, 11, 12, 13, 14, 15, 16, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower boundary:", lower)
print("Upper boundary:", upper)
print("Outliers:", outliers)
```

### How the example demonstrates IQR
Most values are between 10 and 16, while 50 is far away. The calculated upper boundary is used to identify 50 as a potential outlier.

### Business example
Suppose most customer orders are between ₹500 and ₹5,000, but one order is ₹50,000. IQR can flag the unusual order for investigation.

### AI/ML use case
Before training a regression model, IQR can help identify extreme feature values that may strongly influence the model.

## 5. Z-Score Method

A Z-score tells us how many standard deviations a value is from the mean.

**Formula:** `z = (x − mean) / standard deviation`

A common rule is to flag values with `|z| > 3` as potential outliers when the data is approximately normally distributed.

```python
import numpy as np
from scipy.stats import zscore

data = np.array([10, 11, 12, 13, 14, 15, 16, 50])

z_scores = zscore(data)

print("Z-scores:", z_scores)
print("Potential outliers:", data[np.abs(z_scores) > 2])
```

### How the example demonstrates Z-Score
The value 50 is far from the average, so its Z-score has a large absolute value. The code flags values whose absolute Z-score is greater than 2.

### Business example
A bank can calculate Z-scores for transaction amounts. An unusually large transaction can be reviewed.

### AI/ML use case
Z-score screening can be useful for numerical features when the feature is reasonably close to a normal distribution.

### When to be careful
For heavily skewed data, the Z-score method may be less appropriate. IQR or a transformation may work better.

## 6. Box Plot

A **box plot** summarizes a distribution using:
- Minimum/non-outlier lower value
- Q1
- Median
- Q3
- Maximum/non-outlier upper value
- Potential outliers

### Real-world analogy
A box plot is like a compact report card for a dataset: it shows the middle group and unusually distant values.

```python
import matplotlib.pyplot as plt

data = [10, 11, 12, 13, 14, 15, 16, 50]

plt.boxplot(data)
plt.title("Box Plot – Outlier Detection")
plt.ylabel("Value")
plt.show()
```

### Interpretation
The point far above the main box/whiskers represents a potential outlier.

### Business example
A company can use a box plot to inspect employee salaries and investigate unusually high or low values.

### AI/ML use case
Box plots are useful during exploratory data analysis before feature engineering and model training.

## 7. Scatter Plot

A **scatter plot** displays the relationship between two numerical variables.

Outliers can appear as points far away from the main cloud of observations.

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 20]
sales = [10, 15, 18, 25, 28, 32, 5]

plt.scatter(hours, sales)
plt.xlabel("Hours")
plt.ylabel("Sales")
plt.title("Scatter Plot with Possible Outlier")
plt.show()
```

### How the example demonstrates the concept
Most observations form a general pattern, while the point `(20, 5)` is far away from the main pattern.

### Business example
A marketing team can plot advertising spend against sales. A campaign with very high spend but unexpectedly low sales may need investigation.

### AI/ML use case
Scatter plots help identify unusual observations and possible relationships before building regression or classification models.

## 8. Practical Outlier Detection Workflow

1. Understand the business/domain context.
2. Visualize the data.
3. Check for data-entry or measurement errors.
4. Try IQR or Z-score where appropriate.
5. Decide whether to keep, correct, transform, cap, or remove the value.
6. Re-check the model/business result.

**Key lesson:** Do not remove an outlier automatically. It may contain valuable information.

## 1. Histogram

### Purpose
Shows the distribution of a numerical variable by grouping values into bins.

### When to use
Use to understand center, spread, skewness, and possible outliers.

### Advantages
- Easy to understand.
- Good for distribution shape.

### Limitations
- Appearance depends on bin size.
- Exact individual values are not emphasized.

### Real-world analogy
A histogram is like counting how many people fall into each height range.

### Small numerical example
Data: `10, 11, 11, 12, 13, 14, 14, 15`

The histogram groups values into ranges.

### Business example
View distribution of order values.

### AI/ML use case
Inspect feature distributions before preprocessing.

```python
import matplotlib.pyplot as plt

scores = [10, 11, 11, 12, 13, 14, 14, 15]

plt.hist(scores, bins=4)
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

### Interpretation
The bars show how frequently values occur in each range. This helps identify concentration and spread.

## 2. Bar Chart

### Purpose
Compares values across categories.

### When to use
Use for categories such as products, regions, departments, or customer types.

### Advantages
- Excellent for category comparison.
- Simple to communicate.

### Limitations
- Not ideal for continuous distributions.
- Too many categories can become difficult to read.

### Real-world analogy
A bar chart is like comparing the number of books sold by different authors.

### Business example
Compare sales by product category.

### AI/ML use case
Visualize class counts to identify class imbalance.

```python
categories = ["Laptop", "Phone", "Tablet"]
sales = [50, 80, 30]

plt.bar(categories, sales)
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Sales by Product")
plt.show()
```

## 3. Line Chart

### Purpose
Shows changes or trends over an ordered sequence, especially time.

### When to use
Use for daily sales, monthly revenue, temperature, stock measurements, etc.

### Advantages
- Excellent for trends.
- Shows increases and decreases clearly.

### Limitations
- Connecting unrelated categories can be misleading.

### Real-world analogy
A line chart is like watching a journey on a road: it shows how the value changes from one point to the next.

### Business example
Track monthly revenue.

### AI/ML use case
Visualize time-series patterns and monitor model metrics over training iterations.

```python
months = ["Jan", "Feb", "Mar", "Apr"]
revenue = [100, 120, 115, 150]

plt.plot(months, revenue, marker="o")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Monthly Revenue")
plt.show()
```

## 4. Scatter Plot

### Purpose
Shows the relationship between two numerical variables.

### When to use
Use to inspect correlation, clusters, and unusual points.

### Advantages
- Reveals relationships and patterns.
- Helps identify outliers.

### Limitations
- More than two variables require additional visual encodings.
- Correlation does not imply causation.

### Real-world analogy
Plot study hours against marks to see whether more study is associated with higher marks.

### Business example
Advertising spend vs sales.

### AI/ML use case
Inspect feature-target relationships and possible nonlinear patterns.

```python
study_hours = [1, 2, 3, 4, 5]
marks = [50, 55, 65, 75, 85]

plt.scatter(study_hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

## 5. Pie Chart

### Purpose
Shows how categories contribute to a whole.

### When to use
Use for a small number of categories when part-to-whole comparison is important.

### Advantages
- Simple for a few categories.
- Shows proportions visually.

### Limitations
- Difficult to compare similar-sized slices.
- Poor choice for many categories.

### Real-world analogy
A pie is divided into slices; each slice represents a portion of the whole.

### Business example
Show revenue share by product category.

### AI/ML use case
Show class proportions in a small dataset.

```python
labels = ["Product A", "Product B", "Product C"]
values = [50, 30, 20]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Revenue Share")
plt.show()
```

## 6. Density Plot

### Purpose
Shows a smoothed estimate of a numerical distribution.

### When to use
Use when comparing distributions between groups or inspecting shape.

### Advantages
- Smooth view of distribution.
- Useful for comparing groups.

### Limitations
- Smoothing depends on bandwidth.
- Can hide individual observations.

### Real-world analogy
Instead of counting people in fixed height buckets, imagine drawing a smooth curve showing where people are concentrated.

### Business example
Compare customer spending distributions between two customer segments.

### AI/ML use case
Compare distributions of model errors or numerical features.

```python
import seaborn as sns
import matplotlib.pyplot as plt

values = [10, 11, 11, 12, 13, 14, 14, 15, 16, 17]

sns.kdeplot(values, fill=True)
plt.title("Density Plot")
plt.xlabel("Value")
plt.show()
```

## 7. Box Plot

### Purpose
Summarizes a numerical distribution using quartiles and potential outliers.

### When to use
Use for comparing distributions and spotting outliers.

### Advantages
- Compact.
- Good for comparing multiple groups.
- Highlights potential outliers.

### Limitations
- Does not show every data point.
- Distribution shape is less detailed than a histogram.

### Real-world analogy
A box plot is a compact summary of where the middle 50% of observations lie.

### Business example
Compare salaries across departments.

### AI/ML use case
Inspect feature outliers before modeling.

```python
groups = {
    "A": [10, 12, 13, 14, 15],
    "B": [20, 21, 22, 25, 30]
}

plt.boxplot(groups.values(), label=groups.keys())
plt.xlabel("Group")
plt.ylabel("Value")
plt.title("Box Plot")
plt.show()
```

## 8. Violin Plot

### Purpose
Combines a box-plot-like summary with a density shape.

### When to use
Use when you want both distribution shape and summary statistics.

### Advantages
- Shows distribution shape.
- Useful for comparing groups.

### Limitations
- More difficult for beginners.
- Density shape depends on smoothing.

### Real-world analogy
A violin plot is like a box plot with an additional picture of where observations are concentrated.

### Business example
Compare customer spending distributions across customer segments.

### AI/ML use case
Compare prediction errors across different model versions.

```python
data = {
    "Group": ["A"] * 5 + ["B"] * 5,
    "Value": [10, 12, 13, 14, 15, 20, 21, 22, 25, 30]
}

sns.violinplot(data=data, x="Group", y="Value")
plt.title("Violin Plot")
plt.show()
```

## 9. Pair Plot

### Purpose
Shows pairwise relationships among several numerical variables.

### When to use
Use during exploratory data analysis when you have a small number of important features.

### Advantages
- Many relationships in one figure.
- Can reveal clusters and correlations.

### Limitations
- Becomes crowded with many features.
- Can be computationally expensive for large datasets.

### Real-world analogy
A pair plot is like a photo album showing every important pair of variables together.

### Business example
Explore relationships among sales, advertising, website visits, and customers.

### AI/ML use case
Perform initial feature exploration before model building.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "sales": [10, 20, 30, 40, 50],
    "visits": [100, 150, 200, 260, 300],
    "ads": [5, 8, 12, 15, 20]
})

sns.pairplot(df)
plt.show()
```

## 10. Heatmap

### Purpose
Uses color intensity to display values in a matrix.

### When to use
Use for correlation matrices, missing-value patterns, confusion matrices, and other matrix-shaped data.

### Advantages
- Quickly shows strong and weak relationships.
- Good for matrices.

### Limitations
- Too many variables can make it crowded.
- Color interpretation can be misleading if the scale is not clear.

### Real-world analogy
A heatmap is like a weather map: stronger colors indicate stronger values.

### Business example
Display correlation between business metrics.

### AI/ML use case
Visualize feature correlations and model confusion matrices.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "sales": [10, 20, 30, 40, 50],
    "visits": [100, 150, 200, 260, 300],
    "ads": [5, 8, 12, 15, 20]
})

corr = df.corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

## 11. Visualization Selection Guide

| Goal | Recommended visualization |
|---|---|
| Numerical distribution | Histogram / Density Plot |
| Compare categories | Bar Chart |
| Show trend over time | Line Chart |
| Relationship between two numbers | Scatter Plot |
| Part-to-whole with few categories | Pie Chart |
| Detect outliers | Box Plot |
| Distribution shape by group | Violin Plot |
| Explore many pairwise relationships | Pair Plot |
| Show matrix relationships | Heatmap |

### Final lesson
Choose a chart based on the question you want to answer—not just because the chart looks attractive.

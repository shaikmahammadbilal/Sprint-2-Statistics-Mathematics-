## 📈 1. Introduction: Covariance vs Correlation

Both terms help us check if two things change together (e.g., Temperature and Ice Cream sales).

*   **Covariance:** Tells us the **direction** of the relationship (positive or negative). But its value can be anything (like 500 or -0.02), so it is hard to read the exact strength.
*   **Correlation:** A standard version of covariance. It tells us both the **direction and the exact strength**. Its value is always strictly between **-1 and +1**.

---

## 🧭 2. Types of Correlation

### 🟢 A. Positive Correlation (Value near +1)
When one thing increases, the other also increases.
*   **Real-time Example:** As the **summer temperature** goes up in Delhi, **AC electricity bills** also go up.

### 🔴 B. Negative Correlation (Value near -1)
When one thing increases, the other decreases.
*   **Real-time Example:** As you increase your **car speed** on the Mumbai-Pune Expressway, the **travel time** to reach your destination drops.

### 🟡 C. No Correlation (Value near 0)
There is no visible relationship between the two things.
*   **Real-time Example:** Your **shoe size** has absolutely nothing to do with your **college exam marks**.

---

## 🧮 3. Pearson vs Spearman Correlation

*   **Pearson Correlation:** Measures **linear (straight line)** relationships. It works best when data is normal and has no extreme values (outliers).
*   **Spearman Correlation:** Measures **rank or monotonic (ordered)** relationships. It works perfectly even if the data curve bends or contains extreme outliers.

---

## 🛠️ 4. Small & Simple Python Code Examples

Let us create a small dataset of 5 students to see how this works in Python.

```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Simple dataset of 5 students
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Exam_Score': [40, 55, 70, 85, 98],
    'Recreational_Hours': [7, 6, 4, 3, 1]
}
df = pd.DataFrame(data)
df
```

### Calculate Covariance Matrix

```
# Shows the direction of the relationship
df.cov()
```

### Calculate Correlation Matrix (Pearson)

```
# Shows the standard strength (-1 to +1)
df.corr()
```

### Calculate Spearman Rank Correlation

```
# Used for ranked or non-linear data
df.corr(method='spearman')
```

### Plot a Visual Heatmap
A **Heatmap** is a colored grid where colors show how strongly features match.

```
# Short code to draw a correlation heatmap
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.show()
```

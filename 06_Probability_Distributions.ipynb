## 1. Bernoulli Distribution

### Definition
The **Bernoulli distribution** describes a single experiment with only two possible outcomes:

- Success = 1
- Failure = 0

Examples:
- Customer buys / does not buy
- Email is spam / not spam
- Product is defective / not defective

### Formula

P(X = x) = p^x (1-p)^(1-x)

where:
- `p` = probability of success
- `x` = 0 or 1

### Where is it used?
Use Bernoulli distribution when there is **one trial** and only two outcomes.

### Real-world example
A customer either purchases a product or does not purchase it.

### AI/ML example
Binary classification such as:
- Fraud / Not Fraud
- Spam / Not Spam
- Disease / No Disease

### Python Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7

x = np.array([0, 1])
probabilities = np.array([1 - p, p])

plt.bar(x, probabilities)
plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.ylim(0, 1)
plt.show()

print("P(Failure):", 1 - p)
print("P(Success):", p)
```

## 2. Binomial Distribution

### Definition
The **Binomial distribution** gives the probability of getting a certain number of successes in a fixed number of independent Bernoulli trials.

Conditions:
1. Fixed number of trials `n`
2. Each trial has two outcomes
3. Trials are independent
4. Probability of success `p` stays the same

### Formula

P(X = k) = C(n,k) p^k (1-p)^(n-k)

where:
- `n` = number of trials
- `k` = number of successes
- `p` = probability of success

### Numerical example
A coin is tossed 5 times. What is the probability of getting exactly 3 Heads?

n = 5, k = 3, p = 0.5

### Where is it used?
- Quality testing
- Marketing conversion analysis
- Survey results
- A/B testing

### Real-world example
A company sends an offer to 10 customers. Each customer has a 20% chance of buying. We can model the number of buyers with a binomial distribution.

### AI/ML example
Binary classification outcomes over a fixed number of independent observations can be modeled with binomial probabilities.

### Python Visualization

```python
from math import comb

n = 5
p = 0.5

probabilities = []

for k in range(n + 1):
    probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    probabilities.append(probability)

x = np.arange(n + 1)

plt.bar(x, probabilities)
plt.title("Binomial Distribution (n=5, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()

print("Probability of exactly 3 successes:", probabilities[3])
```

### Important difference

**Bernoulli:** one trial.

**Binomial:** multiple independent Bernoulli trials.

Example:
- One customer purchase → Bernoulli
- Number of purchases among 100 customers → Binomial

## 3. Poisson Distribution

### Definition
The **Poisson distribution** models the number of times an event occurs in a fixed interval of time or space when events occur at an average rate.

### Formula

P(X = k) = (e^(-λ) λ^k) / k!

where:
- `λ` (lambda) = average number of events
- `k` = number of events

### Where is it used?
- Number of customer calls per hour
- Website requests per minute
- Number of defects
- Number of arrivals
- Number of accidents in an area

### Real-world example
A support center receives an average of 4 calls per hour. The number of calls in one hour can be modeled using a Poisson distribution if the assumptions are reasonable.

### AI/ML example
Poisson models are useful for count data such as number of events, clicks, incidents, or arrivals.

### Python Visualization

```python
from math import exp, factorial

lam = 4
x = np.arange(0, 13)

probabilities = [
    (exp(-lam) * (lam ** k)) / factorial(k)
    for k in x
]

plt.bar(x, probabilities)
plt.title("Poisson Distribution (lambda=4)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()

print("Probability of exactly 4 events:",
      round(probabilities[4], 4))
```

## 4. Uniform Distribution

### Definition
A **continuous uniform distribution** gives equal probability density to every value inside a specified interval.

For a continuous uniform distribution from `a` to `b`:

### Formula

f(x) = 1 / (b-a), for a ≤ x ≤ b

### Where is it used?
- Random simulation
- Random sampling
- Computer simulations
- Situations where every value in an interval is equally plausible

### Real-world example
Suppose a random arrival time is equally likely to occur anywhere between 10:00 AM and 11:00 AM.

### AI/ML example
Uniform random values are commonly used in simulation and some parameter initialization methods.

### Python Visualization

```python
a = 0
b = 10

uniform_data = np.random.uniform(a, b, 5000)

plt.hist(uniform_data, bins=30, density=True, edgecolor="black")
plt.title("Uniform Distribution (0 to 10)")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()

print("Sample mean:", round(np.mean(uniform_data), 2))
print("Theoretical mean:", (a + b) / 2)
```

## 5. Normal Distribution

### Definition
The **Normal distribution**, also called the **Gaussian distribution**, is a bell-shaped and symmetric continuous distribution.

It is described by:
- Mean `μ`
- Standard deviation `σ`

### Formula

f(x) = [1 / (σ√(2π))] × e^[-(x-μ)^2 / (2σ^2)]

### Where is it used?
- Measurement data
- Test scores
- Manufacturing measurements
- Measurement errors
- Statistical analysis

### Real-world example
Heights of a large population can sometimes be approximately modeled using a normal distribution.

### AI/ML example
Normal distributions are used in Gaussian Naive Bayes, Gaussian mixture models, and models involving Gaussian noise.

### Python Visualization

```python
mu = 70
sigma = 10

normal_data = np.random.normal(mu, sigma, 5000)

plt.hist(normal_data, bins=40, density=True, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()

print("Sample mean:", round(np.mean(normal_data), 2))
print("Sample standard deviation:", round(np.std(normal_data), 2))
```

## 6. Exponential Distribution

### Definition
The **Exponential distribution** models the waiting time until the next event in a Poisson process.

If `λ` is the event rate:

### Formula

f(x) = λe^(-λx), for x ≥ 0

### Where is it used?
- Waiting time between customer arrivals
- Time between website requests
- Time until a failure
- Service waiting times

### Relationship with Poisson distribution

**Poisson:** counts how many events occur in an interval.

**Exponential:** measures the waiting time between events.

### Real-world example
If customers arrive at a service center randomly at an average rate, the waiting time until the next customer can be modeled with an exponential distribution under suitable assumptions.

### AI/ML example
Exponential distributions can be used in survival analysis, reliability modeling, and event-time modeling.

### Python Visualization

```python
rate = 0.5  # average rate per unit of time
scale = 1 / rate

exponential_data = np.random.exponential(scale=scale, size=5000)

plt.hist(exponential_data, bins=40, density=True, edgecolor="black")
plt.title("Exponential Distribution")
plt.xlabel("Waiting Time")
plt.ylabel("Density")
plt.show()

print("Sample mean:", round(np.mean(exponential_data), 2))
print("Theoretical mean:", round(1 / rate, 2))
```

## 8. Important Differences

### Bernoulli vs Binomial
- Bernoulli = one trial.
- Binomial = multiple independent Bernoulli trials.

### Binomial vs Poisson
- Binomial uses a fixed number of trials.
- Poisson counts events in a time/space interval.
- Under certain conditions, Poisson can approximate a binomial distribution when the number of trials is large and the success probability is small.

### Poisson vs Exponential
- Poisson = number of events.
- Exponential = waiting time between events.

### Normal vs Uniform
- Normal = bell-shaped, highest around the mean.
- Uniform = equal density over the interval.

```python
# Quick visual comparison using simulated data

fig, axes = plt.subplots(2, 3, figsize=(14, 8))

# Bernoulli
axes[0, 0].bar([0, 1], [0.3, 0.7])
axes[0, 0].set_title("Bernoulli")

# Binomial
axes[0, 1].bar(x, probabilities if len(probabilities) == len(x) else np.zeros(len(x)))
axes[0, 1].set_title("Binomial")

# Poisson
axes[0, 2].bar(np.arange(0, 13), [
    (exp( -4) * (4 ** k)) / factorial(k) for k in range(13)
])
axes[0, 2].set_title("Poisson")

# Uniform
axes[1, 0].hist(np.random.uniform(0, 10, 2000), bins=25, edgecolor="black")
axes[1, 0].set_title("Uniform")

# Normal
axes[1, 1].hist(np.random.normal(0, 1, 2000), bins=25, edgecolor="black")
axes[1, 1].set_title("Normal")

# Exponential
axes[1, 2].hist(np.random.exponential(2, 2000), bins=25, edgecolor="black")
axes[1, 2].set_title("Exponential")

plt.tight_layout()
plt.show()
```

## 1. Normal Distribution

**Definition:** A normal distribution is a continuous probability distribution with a bell-shaped curve. Most values are near the mean, and fewer values occur as we move away from the mean.

**Main properties:**
- Bell-shaped
- Symmetric around the mean
- Mean = Median = Mode
- Total area under the curve = 1

**Formula:**

$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30)
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

**Numerical Example:** If exam scores have mean = 70 and standard deviation = 10, many students will score close to 70.

**Business Example:** Customer delivery times may approximately follow a normal distribution.

**AI/ML Example:** Some model features may be approximately normally distributed. Standardization is often useful when features have different scales.

**Interpretation:** Values close to the mean are common; values far from the mean are less common.

## 2. Uniform Distribution

**Definition:** In a uniform distribution, every value in a specified interval has the same probability density.

**Formula for continuous uniform distribution:**

$$f(x)=\frac{1}{b-a},\quad a\le x\le b$$

**Example:** A random number between 0 and 10 has equal density throughout the interval.

```python
data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20)
plt.title('Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

**Business Example:** Randomly assigning customers to groups can use uniform random numbers.

**AI/ML Example:** Random initialization or random sampling can use uniform distributions.

**Interpretation:** No part of the interval is favored over another.

## 3. Gaussian Distribution

**Definition:** Gaussian distribution is another name for the normal distribution.

It is described by two parameters:
- **Mean (μ):** center of the distribution
- **Standard deviation (σ):** spread of the distribution

**Business Example:** Measurement errors are often modeled using a Gaussian distribution.

**AI/ML Example:** Gaussian assumptions appear in methods such as Gaussian Naive Bayes and some probabilistic models.

**Interpretation:** A Gaussian distribution is centered around its mean and becomes thinner as values move away from the mean.

## 4. Left Skew (Negative Skew)

**Definition:** A left-skewed distribution has a long tail toward the left side.

**Typical relationship:**

**Mean < Median < Mode**

**Example:** A test where most students score very high, with a few very low scores.

```python
data = np.concatenate([np.random.normal(80, 5, 900), np.random.normal(45, 5, 100)])

plt.hist(data, bins=30)
plt.title('Left-Skewed Distribution (Example)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

**Business Example:** Customer ratings where most ratings are high but a few are very low.

**AI/ML Example:** Skewed features may need transformation before some ML algorithms are applied.

**Interpretation:** The long tail is on the left, and unusually small values pull the mean downward.

## 5. Right Skew (Positive Skew)

**Definition:** A right-skewed distribution has a long tail toward the right side.

**Typical relationship:**

**Mode < Median < Mean**

**Example:** Income data often has a right tail because a small number of people earn very high incomes.

```python
data = np.random.lognormal(mean=3, sigma=0.6, size=1000)

plt.hist(data, bins=30)
plt.title('Right-Skewed Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

**Business Example:** Sales revenue, income, and transaction amounts can be right-skewed.

**AI/ML Example:** Log transformation is commonly considered for strongly right-skewed positive features.

**Interpretation:** A few unusually large values pull the mean toward the right.

## 6. Symmetric Distribution

**Definition:** A distribution is symmetric when its left and right sides have the same overall shape around a center.

**Example:** A normal distribution is symmetric.

**Important:** Symmetry does not automatically mean the distribution is normal.

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.title('Symmetric Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

**Business Example:** A measurement process with balanced variation around a target can be approximately symmetric.

**AI/ML Example:** Understanding symmetry helps decide whether transformations may be useful for a feature.

**Interpretation:** The distribution has similar behavior on both sides of its center.

## 7. Kurtosis

**Definition:** Kurtosis describes the tail behavior of a distribution and, in common statistical usage, helps indicate how much extreme-value behavior exists compared with a normal distribution.

**Types:**
- **Mesokurtic:** Similar tail behavior to the normal distribution.
- **Leptokurtic:** Heavier tails; more extreme values are possible.
- **Platykurtic:** Lighter tails; fewer extreme values are expected.

**Note:** Different software may report either Pearson kurtosis or excess kurtosis. For a normal distribution, Pearson kurtosis is 3 and excess kurtosis is 0.

```python
from scipy.stats import kurtosis

data = np.random.normal(0, 1, 1000)

print('Excess kurtosis:', kurtosis(data))
print('Pearson kurtosis:', kurtosis(data, fisher=False))
```

**Business Example:** In finance, heavy-tailed return distributions can indicate a greater chance of extreme movements.

**AI/ML Example:** Kurtosis can help identify features with unusual tail behavior and potential outliers.

**Interpretation:** Higher excess kurtosis generally means heavier tails and more extreme observations than a normal distribution.

## 8. Empirical Rule (68–95–99.7 Rule)

For a **normal distribution**:

- About **68%** of observations fall within **1 standard deviation** of the mean.
- About **95%** fall within **2 standard deviations**.
- About **99.7%** fall within **3 standard deviations**.

**Formula:**

- 68% → μ ± 1σ
- 95% → μ ± 2σ
- 99.7% → μ ± 3σ

**Numerical Example:** If μ = 100 and σ = 10:
- 68% → 90 to 110
- 95% → 80 to 120
- 99.7% → 70 to 130

```python
mean = 100
std = 10

print('Within 1 SD:', mean - std, 'to', mean + std)
print('Within 2 SD:', mean - 2 * std, 'to', mean + 2 * std)
print('Within 3 SD:', mean - 3 * std, 'to', mean + 3 * std)
```

**Business Example:** A company can use the rule to understand whether delivery times are unusually far from the average when the data is approximately normal.

**AI/ML Example:** Standard deviation ranges are useful for understanding standardized features and detecting unusually distant observations.

**Interpretation:** In a normal distribution, almost all observations are within three standard deviations of the mean.

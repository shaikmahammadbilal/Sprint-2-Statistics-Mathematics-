## 1. Correlation Matrix

### Definition
A correlation matrix shows pairwise correlation between numerical variables. Pearson correlation ranges from -1 to +1.

### Real-world analogy
If umbrella sales increase when rainfall increases, the two variables move together positively.

### Small numerical example
If `study_hours` and `exam_score` have correlation `0.85`, they have a strong positive linear relationship.

### Business example
A retailer can examine relationships among advertising spend, website visits, and sales.

### AI/ML use case
Correlation helps with exploratory analysis and can reveal redundant numerical features.

### Important
Correlation does not prove causation.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5],
    "attendance": [60, 65, 70, 80, 90],
    "score": [50, 55, 65, 75, 85]
})

correlation_matrix = data.corr()
print(correlation_matrix)

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

## 2. Multicollinearity

### Definition
Multicollinearity occurs when two or more predictor variables are strongly related to each other.

### Real-world analogy
Suppose you use both `distance in kilometers` and `distance in meters` as separate model features. They contain almost the same information.

### Small numerical example
- `house_area_sqft`
- `house_area_sqm`

These variables are almost duplicates.

### Business example
A sales model might contain both total advertising spend and a total formed from the same channel spends. This can create redundancy.

### AI/ML use case
Multicollinearity can make linear regression coefficients unstable and harder to interpret.

## 3. Variance Inflation Factor (VIF)

### Definition
VIF measures how much the variance of an estimated regression coefficient is inflated because of relationships among predictors.

A common rule of thumb:
- VIF ≈ 1: little multicollinearity
- VIF above 5: possible concern
- VIF above 10: often considered a strong warning

These are heuristics, not universal laws.

```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = pd.DataFrame({
    "feature_1": [1, 2, 3, 4, 5, 6],
    "feature_2": [2, 4, 6, 8, 10, 12],
    "feature_3": [6, 5, 7, 6, 8, 7]
})

vif = pd.DataFrame()
vif["feature"] = X.columns
vif["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

print(vif)
```

### How VIF demonstrates the concept
`feature_2` is directly related to `feature_1`, so the model has difficulty separating their individual effects. This can produce a high VIF.

### Business example
Before building a pricing regression model, calculate VIF to identify redundant predictors.

### AI/ML use case
For interpretable linear models, reducing high-multicollinearity features can improve coefficient stability.

## 4. Feature Importance

### Definition
Feature importance estimates how useful each feature is for making predictions in a model.

Different algorithms calculate importance differently. Tree models can use split-based importance; permutation importance measures the performance change after shuffling a feature.

### Real-world analogy
For predicting house prices, location may be much more useful than the color of the front door.

### Business example
An e-commerce model predicts whether a customer will purchase. Features such as past purchases, browsing activity, and discount usage may have different importance.

### AI/ML use case
Feature importance helps explain models and prioritize useful variables.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split

X = pd.DataFrame({
    "age": [20, 25, 30, 35, 40, 45, 50, 55],
    "income": [20, 25, 30, 40, 50, 60, 70, 80],
    "visits": [1, 2, 3, 4, 5, 6, 7, 8]
})
y = [0, 0, 0, 0, 1, 1, 1, 1]

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X, y)

importance = pd.Series(model.feature_importances_, index=X.columns)
print(importance.sort_values(ascending=False))
```

## 5. Feature Selection Basics

### Definition
Feature selection means choosing useful features and removing features that are irrelevant, redundant, or harmful.

### Common approaches
- Domain knowledge
- Correlation analysis
- VIF
- Statistical tests
- Recursive feature elimination
- Model-based importance
- L1 regularization

### Real-world analogy
Packing for a trip: take items you need and avoid carrying unnecessary duplicates.

### Business example
A customer-churn model may start with 100 columns and keep only useful, reliable predictors.

### AI/ML use case
Feature selection can reduce training cost, simplify models, reduce noise, and sometimes improve generalization.

## 6. Practical Feature-Relationship Workflow

1. Understand each feature.
2. Check data types and missing values.
3. Inspect correlations.
4. Check multicollinearity when relevant.
5. Use VIF for regression-style analysis.
6. Train a baseline model.
7. Inspect feature importance or permutation importance.
8. Select features using training data only.
9. Validate the final model on unseen data.

### Key lesson
Feature selection should be performed carefully to avoid **data leakage**. Never use information from the test set to make feature-selection decisions.

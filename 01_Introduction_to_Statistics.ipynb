# 1. What is Statistics?

## Definition

**Statistics** is the science of collecting, organizing, analyzing, interpreting, and presenting data.

In simple words:

> Statistics helps us understand data and make decisions from it.

## Simple Example

Suppose the marks of 5 students are:

`60, 70, 80, 90, 100`

We can use statistics to calculate the average mark:

**Average = (60 + 70 + 80 + 90 + 100) / 5 = 80**

So, statistics helps us summarize the data.

## Real-World Example

A company wants to know the average salary of its employees. It collects salary data and calculates the average.

## AI/ML Use Case

Machine Learning models work with data. Statistics helps us understand that data before training a model.

For example, statistics can help identify:
- Average age of customers
- Most common customer category
- Unusual values
- Relationship between variables


```python
marks = [60, 70, 80, 90, 100]

average = sum(marks) / len(marks)

print("Marks:", marks)
print("Average Marks:", average)
```

## Business Example

These statistics can help the company make better decisions.


# 3. Statistics vs Machine Learning

Statistics and Machine Learning are closely related, but their main goals can be different.

| Statistics | Machine Learning |
|---|---|
| Focuses on understanding data | Focuses on learning patterns from data |
| Often emphasizes inference and relationships | Often emphasizes prediction |
| Uses statistical models | Uses algorithms and models |
| Helps estimate population characteristics | Helps make predictions on new data |
| Strong focus on uncertainty and inference | Strong focus on predictive performance |

## Simple Example

Suppose we have house data containing:

- Area
- Number of bedrooms
- Location
- Price

**Statistics:** We may study how area and bedrooms are related to price.

**Machine Learning:** We may train a model to predict the price of a new house.

## Important Point

Statistics and Machine Learning are not completely separate. Modern AI/ML uses many statistical concepts.


# 4. Descriptive Statistics

## Definition

**Descriptive Statistics** describes and summarizes the data we already have.

It does not try to make conclusions about an entire population from a sample.

## Common Methods

- Mean
- Median
- Mode
- Minimum
- Maximum
- Range
- Variance
- Standard deviation
- Tables and graphs

## Numerical Example

Marks:

`60, 70, 80, 90, 100`

The mean is:

`80`

The minimum is:

`60`

The maximum is:

`100`

The range is:

`100 - 60 = 40`

## Real-World Example

A school calculates the average marks of students who took an examination.

That is descriptive statistics because it summarizes the students' collected marks.


```python
marks = [60, 70, 80, 90, 100]

mean = sum(marks) / len(marks)
minimum = min(marks)
maximum = max(marks)
data_range = maximum - minimum

print("Mean:", mean)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", data_range)
```

# 5. Inferential Statistics

## Definition

**Inferential Statistics** uses data from a sample to make conclusions or estimates about a larger population.

## Simple Example

A company has **10,000 customers**.

Instead of asking every customer, the company surveys **500 customers**.

If the sample is properly selected, the company can use the sample results to estimate opinions of the larger customer population.

## Business Example

A company wants to know whether customers like a new product.

- Population = all customers
- Sample = 500 selected customers
- Result = 80% of sampled customers like the product

The company may use this sample to infer customer preference in the larger population.

## AI/ML Connection

Inferential statistics is useful when we want to estimate relationships, test assumptions, or quantify uncertainty from limited data.


# 6. Population

## Definition

A **Population** is the complete group of people, objects, events, or measurements that we are interested in studying.

## Example

A college has 5,000 students.

If we want to study the average height of all students:

**Population = all 5,000 students**

## Business Example

For a bank studying its customers:

**Population = all customers of the bank**

## AI/ML Example

If an ML model is intended to predict purchases for all customers, the target population could be all customers who may use the system.


# 7. Sample

## Definition

A **Sample** is a smaller group selected from a population for study.

## Example

College population:

`5,000 students`

Selected students:

`500 students`

Therefore:

**Population = 5,000 students**

**Sample = 500 students**

## Why Use a Sample?

Studying the entire population may be:
- Expensive
- Time-consuming
- Difficult
- Sometimes impossible

A carefully selected sample can provide useful information about the population.


# 8. Census

## Definition

A **Census** is the process of collecting information from **every member of a population**.

## Example

A company has 1,000 employees and asks every employee about job satisfaction.

This is a census because all 1,000 employees are included.

## Sample vs Census

| Sample | Census |
|---|---|
| Studies part of the population | Studies the entire population |
| Usually cheaper | Usually more expensive |
| Usually faster | Usually slower |
| Can have sampling error | Does not have sampling error from selecting a sample |

## Real-World Example

A national population census attempts to collect information about the entire population of a country.


# 9. Sampling

## Definition

**Sampling** is the process of selecting a sample from a population.

## Example

Population:

`10,000 customers`

Selected sample:

`500 customers`

The process used to select those 500 customers is called sampling.

## Common Sampling Methods

### 1. Simple Random Sampling
Every member has an equal chance of being selected.

### 2. Systematic Sampling
Select every nth member.

### 3. Stratified Sampling
Divide the population into groups and select samples from each group.

### 4. Cluster Sampling
Divide the population into clusters and select some clusters.

## AI/ML Example

When creating a training dataset, careful sampling can help ensure that important groups are properly represented.


# 10. Variables

## Definition

A **Variable** is a characteristic or value that can change from one observation to another.

## Examples

For a student:

- Age
- Gender
- Marks
- Department
- Attendance

These are variables because their values can differ between students.

## Simple Example

| Student | Age | Marks |
|---|---:|---:|

| B | 21 | 82 |

Here:

## AI/ML Connection

Variables are the basic information used to describe observations and can become features used by ML models.


# 11. Features

## Example

- House area
- Number of bedrooms

- House price

## Important

A feature is usually an input to the model, while the **target/label** is what the model is trying to predict.

# 16. Continuous Data

## Definition

**Continuous Data** can take any value within a range and is usually obtained through measurement.

## Examples

- Height = 172.5 cm
- Weight = 68.7 kg
- Temperature = 36.6°C
- Time = 10.25 seconds

For example, height could be:

`170 cm`, `170.5 cm`, `170.55 cm`, etc.

## Business Example

A manufacturing company measures the weight of products. Product weight can have decimal values.

## AI/ML Use Case

Continuous measurements are commonly used as numerical features in ML models.


# 17. Discrete Data

## Definition

**Discrete Data** consists of countable values, usually whole numbers.

## Examples

- Number of students = 50
- Number of products = 10
- Number of employees = 25
- Number of website visits = 100

You cannot normally have:

`3.5 students`

Therefore, number of students is discrete data.

## Continuous vs Discrete

| Continuous | Discrete |
|---|---|
| Measured | Counted |
| Can have decimal values | Usually whole/count values |
| Height | Number of students |
| Weight | Number of products |
| Temperature | Number of orders |


# 18. Structured Data

## Definition

**Structured Data** is data organized in a fixed format, usually rows and columns.

## Example

| Employee_ID | Name | Department | Salary |
|---:|---|---|---:|
| 101 | Ravi | IT | 50000 |
| 102 | Sita | HR | 45000 |
| 103 | John | Sales | 48000 |

This is structured data because it has a clear schema and columns.

## Common Examples

- SQL tables
- Excel spreadsheets
- CSV files
- Relational databases

## AI/ML Use Case

Many traditional ML models work with structured tabular datasets containing rows and columns.

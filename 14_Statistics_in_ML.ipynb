## 1. Statistics in Data Cleaning

### Definition
Data cleaning means finding and fixing problems such as incorrect values, duplicates, inconsistent formats, and impossible observations.

### Real-world analogy
Before cooking, you wash vegetables and remove spoiled pieces. Data cleaning prepares raw data before modeling.

### Small numerical example
Ages:

`20, 22, 25, 24, 200`

Age 200 is probably invalid for a normal customer dataset.

### Business example
A bank checks customer records for impossible ages, duplicate accounts, and invalid transaction amounts.

### AI/ML use case
Cleaning incorrect training data can prevent a model from learning misleading patterns.

### How the example demonstrates the concept
The value 200 should be investigated rather than blindly used as a real customer age.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "age": [20, 22, 25, 24, 200],
    "income": [25000, 30000, 35000, 32000, 40000]
})

print("Before cleaning:")
print(df)

# Flag suspicious ages
df["age_valid"] = df["age"].between(0, 100)

print("\nValidation result:")
print(df)
```

## 2. Statistics in Missing Value Handling

### Definition
Missing value handling means deciding what to do when observations are unavailable.

Common approaches:
- Remove rows/columns when appropriate.
- Fill numerical values with mean or median.
- Fill categorical values with mode.
- Use more advanced imputation methods.

### Real-world analogy
If one student's mark is missing, you might use information from other students to estimate it rather than leaving the record unusable.

### Small numerical example
Salaries:

`30000, 32000, NaN, 35000, 34000`

The median can be used as a simple robust replacement.

### Business example
Customer income may be missing in a loan dataset.

### AI/ML use case
Most ML algorithms require a defined value for each feature, so missing-value handling is often necessary.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "salary": [30000, 32000, np.nan, 35000, 34000]
})

median_salary = df["salary"].median()
df["salary_filled"] = df["salary"].fillna(median_salary)

print(df)
```

## 3. Statistics in Outlier Detection

### Definition
Statistics helps identify observations that are unusually far from the main distribution.

Common methods:
- IQR
- Z-score
- Robust statistics
- Visualization

### Real-world analogy
If almost every delivery takes 20–40 minutes but one record says 900 minutes, investigate it.

### Business example
Detect unusually large bank transactions.

### AI/ML use case
Extreme feature values can influence some models strongly, especially linear models.

### Example
Data: `10, 11, 12, 13, 14, 50`

The value 50 is a candidate outlier.

```python
import numpy as np

data = np.array([10, 11, 12, 13, 14, 50])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Potential outliers:", data[(data < lower) | (data > upper)])
```

## 4. Statistics in Feature Engineering

### Definition
Feature engineering creates useful input variables from raw data.

Statistics can help with:
- Aggregation
- Ratios
- Counts
- Rolling averages
- Log transformations
- Binning
- Interaction features

### Real-world analogy
Raw ingredients are transformed into a useful meal. Raw data can be transformed into useful model features.

### Small numerical example
Raw:
- Total purchases = 10
- Customer age = 25

Create:
- Average purchases per month
- Purchase frequency
- Age group

### Business example
An e-commerce company creates `average_order_value = total_revenue / number_of_orders`.

### AI/ML use case
Good features can make patterns easier for ML models to learn.

```python
import pandas as pd

df = pd.DataFrame({
    "customer": ["A", "B", "C"],
    "revenue": [1000, 1500, 800],
    "orders": [10, 15, 4]
})

df["average_order_value"] = df["revenue"] / df["orders"]

print(df)
```

## 5. Statistics in Feature Scaling

### Definition
Feature scaling puts numerical features on comparable scales.

Common methods:
- Standardization
- Min-Max scaling
- Robust scaling

### Real-world analogy
Imagine comparing distance in kilometers with distance in millimeters. Scaling makes the numbers easier to compare.

### Small numerical example
Feature A: `10, 20, 30`  
Feature B: `1000, 2000, 3000`

Feature B has much larger numeric values.

### Business example
Customer age and annual income may have very different scales.

### AI/ML use case
Scaling is especially important for distance-based and gradient-based algorithms such as KNN, K-Means, and many neural-network workflows.

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

X = np.array([
    [20, 30000],
    [30, 50000],
    [40, 70000]
])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled)
```

## 6. Statistics in Data Normalization

### Definition
Normalization often refers to transforming data so values fit a desired scale, commonly 0 to 1 with Min-Max scaling.

### Formula
`x_scaled = (x - min) / (max - min)`

### Small numerical example
For values 10, 20, 30:

For 20:

`(20 - 10) / (30 - 10) = 0.5`

### Real-world analogy
Converting different-sized measurements to the same 0-to-1 measuring scale.

### Business example
Normalize customer activity scores before combining them into a composite score.

### AI/ML use case
Useful for algorithms sensitive to feature magnitude.

### Important
"Normalization" can mean different things in different ML contexts, so always check the exact transformation being used.

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

data = np.array([[10], [20], [30]])

scaler = MinMaxScaler()
normalized = scaler.fit_transform(data)

print(normalized)
```

## 7. Statistics in Model Evaluation

### Definition
Statistics helps us measure how well a model performs on unseen data.

Common metrics:
- Regression: MAE, MSE, RMSE, R²
- Classification: accuracy, precision, recall, F1-score, ROC-AUC
- Cross-validation for estimating generalization performance

### Real-world analogy
A student should be tested on questions they did not memorize from the practice sheet.

### Business example
A fraud model must be evaluated on transactions not used during training.

### AI/ML use case
Use train/validation/test splits and cross-validation to estimate performance and avoid overfitting.

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris

X, y = load_iris(return_X_y=True)

model = DecisionTreeClassifier(random_state=42)
scores = cross_val_score(model, X, y, cv=5)

print("Cross-validation scores:", scores)
print("Mean score:", scores.mean())
```

## 8. Linear Regression

### Definition
Linear regression models a numerical target as a linear function of one or more input features.

genui{"learning_viz":{"type_id":"LEAST_SQUARE_REGRESSION"}}

### Real-world analogy
Draw the straight line that best summarizes how house price changes with house size.

### Small numerical example
If:

`price = 50 + 10 × size`

then for `size = 5`:

`price = 50 + 10(5) = 100`

### Business example
Predict sales from advertising spend.

### AI/ML use case
Regression is used for predicting continuous values such as revenue, demand, price, and delivery time.

### Statistics connection
Linear regression estimates coefficients and commonly uses concepts such as residuals, variance, least squares, confidence intervals, and hypothesis tests.

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([10, 20, 30, 40, 50])

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Prediction for 6:", model.predict([[6]])[0])
```

## 9. Logistic Regression

### Definition
Logistic regression is commonly used for classification. It estimates probabilities and maps them to classes using a decision rule.

### Real-world analogy
A loan system estimates the probability that a customer will default and then uses a threshold to make a decision.

### Small numerical example
Suppose a model predicts:

`P(default) = 0.80`

If the decision threshold is 0.50, classify the customer as likely to default.

### Business example
Predict whether a customer will churn.

### AI/ML use case
Binary classification such as spam/not spam, fraud/not fraud, churn/not churn.

### Statistics connection
Logistic regression models the log-odds of the outcome using input features.

```python
from sklearn.linear_model import LogisticRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

probabilities = model.predict_proba([[4], [5]])
print("Class probabilities:")
print(probabilities)

print("Predicted classes:", model.predict([[4], [5]]))
```

## 10. Naive Bayes

### Definition
Naive Bayes is a probabilistic classification algorithm based on Bayes' theorem and a conditional-independence assumption.

genui{"learning_viz":{"type_id":"BAYES_THEOREM"}}

### Real-world analogy
Suppose you want to decide whether an email is spam. Words such as "offer", "free", and "winner" provide evidence that updates the probability of spam.

### Small numerical example
If an email contains words strongly associated with spam, the estimated probability of spam increases.

### Business example
Classify customer messages into categories.

### AI/ML use case
Spam detection, document classification, sentiment/topic classification.

### Statistics connection
The model uses probabilities and Bayes' theorem to update beliefs based on evidence.

```python
from sklearn.naive_bayes import GaussianNB
import numpy as np

X = np.array([
    [1, 10],
    [2, 12],
    [8, 30],
    [9, 32]
])
y = np.array([0, 0, 1, 1])

model = GaussianNB()
model.fit(X, y)

print("Prediction:", model.predict([[3, 15]]))
print("Probabilities:", model.predict_proba([[3, 15]]))
```

## 11. Clustering

### Definition
Clustering groups similar observations without predefined labels.

### Real-world analogy
A supermarket groups customers with similar shopping behavior even when nobody has labeled the customers beforehand.

### Small numerical example
Customers:
- A: spends 100, visits 2 times
- B: spends 110, visits 2 times
- C: spends 900, visits 20 times

A clustering algorithm may place A and B together and C in another cluster.

### Business example
Customer segmentation.

### AI/ML use case
K-Means, hierarchical clustering, and DBSCAN can discover groups in unlabeled data.

### Statistics connection
Distance, variance, distributions, and cluster summaries help describe and evaluate groups.

```python
from sklearn.cluster import KMeans
import numpy as np

X = np.array([
    [100, 2],
    [110, 2],
    [120, 3],
    [900, 20],
    [850, 18],
    [950, 22]
])

model = KMeans(n_clusters=2, random_state=42, n_init=10)
labels = model.fit_predict(X)

print("Cluster labels:", labels)
print("Cluster centers:")
print(model.cluster_centers_)
```

## 12. Principal Component Analysis (PCA)

### Definition
PCA transforms correlated numerical features into a smaller set of new features called principal components.

### Real-world analogy
Imagine having 10 photos that contain nearly the same information. PCA tries to summarize the important variation using fewer dimensions.

### Small numerical example
Suppose height and weight are strongly related. PCA can create a component that captures much of their shared variation.

### Business example
Reduce hundreds of correlated customer attributes before analysis.

### AI/ML use case
Dimensionality reduction, visualization, noise reduction, and preprocessing.

### Statistics connection
PCA is based on variance and covariance structure.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import numpy as np

X = np.array([
    [170, 65],
    [175, 70],
    [180, 75],
    [185, 80],
    [190, 85]
])

X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X_scaled)

print("Reduced data:")
print(X_reduced)
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

## 13. Recommendation Systems

### Definition
Recommendation systems suggest products, movies, songs, or content that a user may like.

### Real-world analogy
A shop assistant remembers what you liked before and suggests similar products.

### Small numerical example
User A rates:
- Movie 1 = 5
- Movie 2 = 4
- Movie 3 = 1

User B has similar ratings. A recommendation system may use that similarity to recommend Movie 2 or another movie that User B liked.

### Business example
An e-commerce website recommends products based on customer behavior.

### AI/ML use case
Netflix-style movie recommendations, shopping recommendations, music recommendations, and content ranking.

### Statistics connection
Recommendation systems use concepts such as averages, similarity, probability, matrix factorization, distributions, and evaluation metrics.

```python
import numpy as np

# Rows = users, columns = products
ratings = np.array([
    [5, 4, 1],
    [5, 4, 2],
    [1, 2, 5]
])

# Simple item popularity example
average_ratings = ratings.mean(axis=0)

print("Average rating for each product:", average_ratings)
print("Most popular product index:", np.argmax(average_ratings) + 1)
```

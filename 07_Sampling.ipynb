# Notebook 7 – Sampling (07_Sampling.ipynb)

Sampling means selecting a small group (sample) from a large group (population) to understand something about the whole population. It saves a lot of time, money, and effort.

## 📋 Basic Terms

* **Population:** The entire group you want to study. 
  * *Real-Time Example:* All users of a food delivery app in India (like Zomato or Swiggy).
* **Sample:** The small part of the group you actually pick to collect data from.
  * *Real-Time Example:* 1,000 users picked from the app to ask for feedback.
* **Sampling Frame:** The actual list from which you pick your sample.
  * *Real-Time Example:* The database list of active app users who ordered food in the last 30 days.

## 🎲 Types of Sampling with Code Examples

First, let us create a simple dataset of 12 employees in an Indian company with their departments (`IT` or `HR`) to use in our examples.

```python
import pandas as pd
import numpy as np

# Simple dataset of 12 employees
data = {
    'ID': range(1, 13),
    'Dept': ['IT', 'IT', 'IT', 'IT', 'IT', 'IT', 'HR', 'HR', 'HR', 'HR', 'HR', 'HR'],
    'Salary': [60, 65, 70, 75, 80, 85, 40, 45, 50, 55, 58, 62]
}
df = pd.DataFrame(data)
print(df)
```

### 1. Simple Random Sampling
Every member has an **equal chance** of getting selected. It is like drawing names out of a lucky draw bowl.

* **Real-Time Example:** A school picks 3 random students from a box containing all student names to win a prize.
* **Code Example:** Select 3 random employees from our dataset.

```python
# Pick 3 random rows
df.sample(n=3, random_state=42)
```

### 2. Stratified Sampling
You divide the population into subgroups (**strata**) based on a trait (like department or age), and then pick randomly from *each* subgroup. This ensures all groups are fairly represented.

* **Real-Time Example:** A college wants feedback on a new rule, so they randomly pick 50 students from the Engineering branch, 50 from Commerce, and 50 from Science.
* **Code Example:** Pick 2 random employees from each department (`IT` and `HR`).

```python
# Pick 2 random rows from each group
df.groupby('Dept').sample(n=2, random_state=42)
```

### 3. Systematic Sampling
You arrange the list in order, pick a random starting point, and then pick every k-th member (like every 3rd or 5th person).

* **Real-Time Example:** Security guards at a metro station check every 10th passenger entering the station.
* **Code Example:** Select every 3rd employee from our list.

```python
# Select every 3rd row starting from index 0
df.iloc[::3]
```

### 4. Cluster Sampling
You divide the population into mixed groups called **clusters** (like cities or buildings). You pick a few entire clusters randomly and study everyone inside them.

* **Real-Time Example:** A telecom company wants to test a new 5G network in a state. Instead of visiting all cities, they randomly select 3 cities and test all users in those 3 cities.
* **Code Example:** Let us treat `Dept` as clusters and randomly select one whole cluster/department.

```python
# Randomly pick one department cluster ('IT' or 'HR') and take all its rows
chosen_cluster = np.random.choice(df['Dept'].unique())
df[df['Dept'] == chosen_cluster]
```

## ⚠️ Biases, Errors, and Sample Size

### 1. Sampling Bias
This happens when some members are systematically more or less likely to get selected. This makes the sample unfair.
* **Real-Time Example:** If you conduct an online survey about internet connectivity issues in India, you automatically miss out on rural people who do not have an internet connection. This is a bias.

### 2. Sampling Error
This is the natural difference between your sample result and the true population result. It happens purely by chance because a sample is just a part of the whole.
* **Important Tip:** You can easily **reduce sampling error by increasing your sample size** (studying more people).

### 3. Sample Size
Choosing the right size is very important. If it is too small, your results will not be accurate. If it is too big, you waste time and money.
* **Real-Time Example:** Predicting national election results by talking to only 10 voters will give a wrong idea (sample size too small). Talking to 50,000 voters across different regions will give a much better prediction.
